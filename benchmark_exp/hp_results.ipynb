```python
import pandas as pd
from aeon.visualisation import plot_critical_difference
```

```python
MODEL="KNN"
mode=""

results = pd.read_csv(f"eval/HP_tuning/uni/{MODEL}{mode}.csv")

all_hps = results["HP"].unique()
print(f"All hyperparameters: {all_hps}")

all_metrics = results.columns[2:]
print(f"All metrics: {all_metrics.tolist()}")

current_metric = "VUS-PR"
assert current_metric in all_metrics, f"Metric {current_metric} not found in results."

current_results = results[["file", "HP", current_metric]]

current_results = current_results.pivot(index="file", columns="HP", values=current_metric)

plot = plot_critical_difference(current_results.values, current_results.columns)
```
